# VGGNet.py

```python
import torch
from VGGNet import VGGBlock

block = VGGBlock(in_channels=64, out_channels=128, num_convs=2)

test_input = torch.randn(2, 64, 150, 150)
output = block(test_input)

print(output.shape)
```

### Meaning:

*  2: two images in the test batch.
*  128: the block changed 64 channels into 128 channels.
*  75, 75: max pooling reduced 150×150 to 75×75.

Once that works, the next block is the main VGGNet class, where we connect an initial convolution, two VGG blocks, pooling, flattening, and the final classifier.

```python
import importlib
import VGGNet as vgg_module

importlib.reload(vgg_module)

model = vgg_module.VGGNet(num_classes=18)
```

```python
import torch

test_input = torch.randn(2, 3, 150, 150)
output = model(test_input)

print(output.shape)
```

```python
import torch.nn as nn

model.train()

test_input = torch.randn(2, 3, 150, 150)
labels = torch.tensor([0, 5])

outputs = model(test_input)

criterion = nn.CrossEntropyLoss()
loss = criterion(outputs, labels)

loss.backward()

print("Loss:", loss.item())
print("Backpropagation completed successfully.")
```

That passed. A loss around 2.87 is normal for random input and random labels; the important result is that loss.backward() completed without error.
